#### Описание данных: Предположим, что у нас есть набор данных в формате CSV, содержащий информацию о производственном процессе. Структура данных может включать следующие поля:

3. Написание программы MapReduce:

В этом примере мы хотим подсчитать общее время работы для каждой машины и количество операций, которые завершились успешно.

```python
Mapper:
import java.io.IOException;
import org.apache.hadoop.io.LongWritable;
import org.apache.hadoop.io.Text;
import org.apache.hadoop.mapreduce.Mapper;

public class ProductionMapper extends Mapper {
  public void map(LongWritable key, Text value, Context context) throws IOException, InterruptedException {
    String[] fields = value.toString().split(",");
    String machineId = fields[1].trim();
    String operationTime = fields[2].trim();
    String status = fields[3].trim();

    context.write(new Text(machineId), new Text(operationTime + "," + status));
  }
}

Reducer:
import java.io.IOException;
import org.apache.hadoop.io.Text;
import org.apache.hadoop.mapreduce.Reducer;

public class ProductionReducer extends Reducer {
  public void reduce(Text key, Iterable values, Context context) throws IOException, InterruptedException {
    int totalOperationTime = 0;
    int completedCount = 0;

    for (Text value : values) {
      String[] fields = value.toString().split(",");
      totalOperationTime += Integer.parseInt(fields[0]);
      if (fields[1].equals("Completed")) {
        completedCount++;
      }
    }

    context.write(key, new Text("Total Operation Time: " + totalOperationTime + ", Completed Count: " + completedCount));
  }
}
```

Уравнение Беллмана и условие оптимальности — это ключевые концепции в теории динамического программирования, которые позволяют находить оптимальные решения для задач с многошаговыми процессами принятия решений. Эти идеи были разработаны Ричардом Беллманом в 1950-х годах и широко применяются в таких областях, как экономика, управление, искусственный интеллект, операционные исследования и другие.

---

### **1. Уравнение Беллмана**

Уравнение Беллмана выражает принцип оптимальности, который гласит: "Оптимальная стратегия обладает тем свойством, что каковы бы ни были начальное состояние и начальное решение, оставшиеся решения должны составлять оптимальную стратегию относительно состояния, возникающего после первого шага."

#### Математическая формулировка:
Пусть $ V(s) $ — это функция стоимости (или значение), которая представляет собой максимальную (или минимальную) стоимость, которую можно получить, начиная из состояния $ s $. Тогда уравнение Беллмана записывается следующим образом:

$$
V(s) = \max_{a} \left[ R(s, a) + \gamma \sum_{s'} P(s' | s, a) V(s') \right],
$$

где:
- $ s $ — текущее состояние системы;
- $ a $ — действие, которое можно предпринять в состоянии $ s $;
- $ R(s, a) $ — немедленная награда (или стоимость), получаемая при выполнении действия $ a $ в состоянии $ s $;
- $ P(s' | s, a) $ — вероятность перехода из состояния $ s $ в состояние $ s' $ при выполнении действия $ a $;
- $ \gamma \in [0, 1] $ — коэффициент дисконтирования, который определяет важность будущих наград по сравнению с текущими;
- $ V(s') $ — значение функции стоимости для нового состояния $ s' $.

#### Интерпретация:
- Первый член $ R(s, a) $ отражает немедленную награду за действие $ a $ в состоянии $ s $.
- Второй член $ \gamma \sum_{s'} P(s' | s, a) V(s') $ учитывает ожидаемую будущую награду, дисконтированную на коэффициент $ \gamma $, при переходе в новое состояние $ s' $.

Уравнение Беллмана рекурсивно связывает значение текущего состояния $ V(s) $ со значениями возможных будущих состояний $ V(s') $.

---

### **2. Условие оптимальности**

Условие оптимальности в контексте уравнения Беллмана означает, что оптимальная стратегия (политика) должна максимизировать (или минимизировать) общую награду, начиная с любого состояния. Это достигается путем выбора такого действия $ a^* $, которое обеспечивает максимальное значение правой части уравнения Беллмана:

$$
a^*(s) = \arg\max_{a} \left[ R(s, a) + \gamma \sum_{s'} P(s' | s, a) V(s') \right].
$$

Здесь:
- $ a^*(s) $ — оптимальное действие в состоянии $ s $;
- $ \arg\max $ означает, что мы выбираем действие $ a $, которое максимизирует выражение.

Таким образом, оптимальная политика $ \pi^* $ определяется как:

$$
\pi^*(s) = a^*(s).
$$

#### Принцип оптимальности:
Принцип оптимальности Беллмана утверждает, что если мы знаем оптимальное значение $ V(s) $ для всех состояний $ s $, то мы можем восстановить оптимальную политику $ \pi^* $, выбирая на каждом шаге действие, которое максимизирует ожидаемую награду.

### **3. Пример применения**

Рассмотрим простой пример управления запасами товаров в магазине:
- Состояние $ s $ — количество товаров на складе.
- Действие $ a $ — заказать дополнительное количество товаров.
- Награда $ R(s, a) $ — прибыль от продаж минус затраты на хранение и заказ.
- Переходы $ P(s' | s, a) $ — вероятность изменения запасов в зависимости от спроса.

Используя уравнение Беллмана, можно найти оптимальную стратегию заказа товаров, которая максимизирует долгосрочную прибыль.

### **4. Численные методы решения**

Уравнение Беллмана обычно решается численно, так как аналитическое решение возможно только для очень простых случаев. Основные методы:
1. **Итерация по значениям (Value Iteration):**
   - Начинаем с произвольной функции $ V_0(s) $.
   - Итеративно обновляем значения:
     $$
     V_{k+1}(s) = \max_{a} \left[ R(s, a) + \gamma \sum_{s'} P(s' | s, a) V_k(s') \right].
     $$
   - Процесс продолжается до сходимости $ V_k(s) \to V^*(s) $.

2. **Итерация по политикам (Policy Iteration):**
   - Начинаем с произвольной политики $ \pi_0(s) $.
   - На каждом шаге:
     - Вычисляем значения $ V(s) $ для текущей политики (оценка политики).
     - Улучшаем политику, выбирая действия, которые максимизируют $ V(s) $ (улучшение политики).
   - Повторяем до сходимости.
